```python
"""
Dataset download and preparation script.
MNIST, CIFAR-10, CIFAR-100 are downloaded automatically via torchvision.
ImageNet must be placed manually in data/imagenet/.
CUB-200-2011 is downloaded via this script.
Stanford Cars is downloaded via torchvision (>=0.15).
"""
```

```python
import os
import tarfile
import urllib.request
import torchvision
import torchvision.transforms as transforms
from torchvision import datasets
```

```python
DATA_ROOT = os.path.join(os.getcwd(), "data")
os.makedirs(DATA_ROOT, exist_ok=True)
```

```python
def download_mnist():
    print("Downloading MNIST")
    datasets.MNIST(root=os.path.join(DATA_ROOT, "mnist"), train=True, download=True)
    datasets.MNIST(root=os.path.join(DATA_ROOT, "mnist"), train=False, download=True)
    print("MNIST OK.")

def download_cifar10():
    print("Downloading CIFAR-10")
    datasets.CIFAR10(root=os.path.join(DATA_ROOT, "cifar10"), train=True, download=True)
    datasets.CIFAR10(root=os.path.join(DATA_ROOT, "cifar10"), train=False, download=True)
    print("CIFAR-10 OK.")

def download_cifar100():
    print("Downloading CIFAR-100")
    datasets.CIFAR100(root=os.path.join(DATA_ROOT, "cifar100"), train=True, download=True)
    datasets.CIFAR100(root=os.path.join(DATA_ROOT, "cifar100"), train=False, download=True)
    print("CIFAR-100 OK.")

def download_cub200():
    cub_dir = os.path.join(DATA_ROOT, "cub200")
    os.makedirs(cub_dir, exist_ok=True)
    tgz_path = os.path.join(cub_dir, "CUB_200_2011.tgz")

    if not os.path.exists(os.path.join(cub_dir, "CUB_200_2011")):
        print("Downloading CUB-200-2011")
        url = "https://data.caltech.edu/records/65de6-vp158/files/CUB_200_2011.tgz"
        urllib.request.urlretrieve(url, tgz_path)
        print("Extracting CUB-200-2011")
        with tarfile.open(tgz_path, "r:gz") as tar:
            tar.extractall(path=cub_dir)
        print("CUB-200-2011 OK.")
    else:
        print("CUB-200-2011 already exists.")

def download_stanford_cars():
    cars_dir = os.path.join(DATA_ROOT, "cars")
    os.makedirs(cars_dir, exist_ok=True)
    print("Downloading Stanford Cars")
    try:
        datasets.StanfordCars(root=cars_dir, split="train", download=True)
        datasets.StanfordCars(root=cars_dir, split="test", download=True)
        print("Stanford Cars OK.")
    except Exception as e:
        print(f"Stanford Cars error: {e}")
        print("Please download manually from https://ai.stanford.edu/~jkrause/cars/car_dataset.html")

def check_imagenet():
    imagenet_dir = os.path.join(DATA_ROOT, "imagenet")
    train_dir = os.path.join(imagenet_dir, "train")
    val_dir = os.path.join(imagenet_dir, "val")
    if not os.path.isdir(train_dir) or not os.path.isdir(val_dir):
        print("WARNING: ImageNet not found in data/imagenet/train and data/imagenet/val.")
        print("Please download manually from https://image-net.org/")
    else:
        print("ImageNet found OK.")
```
